### 26. Determinant and inverse  <span style='color:#C92A2A;font-weight:bold'>[HARD]</span>

Find the determinant and inverse of `[[4, 7], [2, 6]]`. Check that A @ inverse gives the identity.

```python
a=np.array([[4,7],[2,6]])
print(np.linalg.det(a))
invA=np.linalg.inv(a)
print(invA)
```

```python
print(a@invA)            #look at the difference these codes
print(np.round(a@invA))
```
